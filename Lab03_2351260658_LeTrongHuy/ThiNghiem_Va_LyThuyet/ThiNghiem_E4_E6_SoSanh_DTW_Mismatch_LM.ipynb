## 3. THÍ NGHIỆM E6: THAY ĐỔI QUYẾT ĐỊNH KHI ĐIỀU CHỈNH TRỌNG SỐ $\lambda_{LM}$
### Mục tiêu:
Khảo sát sự cân bằng giữa điểm âm học và điểm ngôn ngữ trong công thức MAP Re-ranking:
$$\text{Score} = \frac{\text{Acoustic Score}}{T} + \lambda_{LM} \cdot \log P(w \mid w_{prev})$$
