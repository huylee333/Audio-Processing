### Phân tích Thí nghiệm E2:

- **39 MFCC động ($13 + 13\Delta + 13\Delta^2$):** Bổ sung thông tin động học thời gian liên khung, giúp phân biệt rõ các âm có phổ tĩnh tương đồng nhưng hướng biến thiên khác nhau.

## 3. THÍ NGHIỆM E3: DIAGONAL SO VỚI FULL COVARIANCE MATRIX
### Mục tiêu:
Khảo sát sự suy biến ma trận hiệp phương sai khi kích thước tập huấn luyện hữu hạn.


### Bảng kết quả Thí nghiệm E3:
| Loại Covariance | Số tham số hiệp phương sai / trạng thái | Accuracy | Hiện tượng số học |
|:---|:---:|:---:|:---|
| **Diagonal (`diag`)** | $D = 39$ | **$100.00\%$** | Hội tụ ổn định tuyệt đối, không suy biến ma trận |
| **Full (`full`)** | $\frac{D(D+1)}{2} = 780$ | **$60.00\%$** | Ma trận suy biến (Singular), thiếu dữ liệu ước lượng, overfit nặng |

### Kết luận khoa học:
Với tập dữ liệu quy mô nhỏ, ma trận hiệp phương sai đường chéo (`diag`) kết hợp variance flooring là lựa chọn tối ưu bắt buộc.
