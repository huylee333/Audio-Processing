Cell 1 - Chuyá»ƒn Mp3 sang wav 16 bits

```python
from pathlib import Path
import shutil
import wave

# File đầu vào
input_file = Path("audio/Thien_ly_oi.mp3")

# File WAV đầu ra
output_file = Path("audio/input.wav")
output_file.parent.mkdir(exist_ok=True)


def get_wav_info(path):
    try:
        with wave.open(str(path), "rb") as wav:
            Fs = wav.getframerate()
            channels = wav.getnchannels()
            sample_width = wav.getsampwidth()
            nframes = wav.getnframes()
            duration = nframes / Fs
        return Fs, channels, sample_width, nframes, duration
    except (EOFError, wave.Error):
        return None


wav_info = get_wav_info(output_file) if output_file.exists() else None

if wav_info is not None:
    print(f"Đã có file WAV hợp lệ: {output_file}")
else:
    if output_file.exists():
        print(f"File WAV hiện có không hợp lệ, sẽ tạo lại: {output_file}")

    if not input_file.exists():
        raise FileNotFoundError(f"Không tìm thấy file MP3: {input_file}")

    try:
        import soundfile as sf

        audio_data, source_fs = sf.read(input_file, always_2d=False)
        sf.write(output_file, audio_data, source_fs, subtype="PCM_16")
        print("Đã chuyển đổi MP3 sang WAV 16-bit bằng soundfile.")
    except Exception as soundfile_error:
        ffmpeg_path = shutil.which("ffmpeg") or shutil.which("avconv")
        ffprobe_path = shutil.which("ffprobe") or shutil.which("avprobe")
        if ffmpeg_path is None or ffprobe_path is None:
            raise RuntimeError(
                "Không thể đọc MP3 bằng soundfile và cũng không tìm thấy ffmpeg/ffprobe. "
                "Hãy cài FFmpeg hoặc đặt sẵn file audio/input.wav hợp lệ."
            ) from soundfile_error

        try:
            from pydub import AudioSegment
        except ModuleNotFoundError as exc:
            raise ModuleNotFoundError(
                "Thiếu thư viện pydub. Cài bằng: pip install pydub"
            ) from exc

        audio = AudioSegment.from_file(input_file, format="mp3")
        audio = audio.set_sample_width(2)
        audio.export(output_file, format="wav")
        print("Đã chuyển đổi MP3 sang WAV 16-bit bằng pydub.")

    wav_info = get_wav_info(output_file)
    if wav_info is None:
        raise RuntimeError(f"Không tạo được file WAV hợp lệ: {output_file}")

Fs, channels, sample_width, nframes, duration = wav_info

print(f"File WAV: {output_file}")
print("\n===== THÔNG TIN FILE WAV =====")
print(f"Sampling rate : {Fs} Hz")
print(f"Channels      : {channels}")
print(f"Sample width  : {sample_width} bytes")
print(f"Bit depth     : {sample_width * 8} bits")
print(f"Duration      : {duration:.3f} s")
print(f"File size     : {output_file.stat().st_size / 1024:.2f} KB")
```

```python
# CELL 02 - Import thÆ° viá»‡n

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import signal
from scipy.io import wavfile
import soundfile as sf

from IPython.display import Audio, display

# ThÆ° má»¥c
AUDIO_DIR = Path("audio")
FIGURE_DIR = Path("figures")

AUDIO_DIR.mkdir(exist_ok=True)
FIGURE_DIR.mkdir(exist_ok=True)

# File WAV Ä‘Ã£ chuyá»ƒn Ä‘á»•i
WAV_FILE = AUDIO_DIR / "input.wav"

print("File Ä‘ang xá»­ lÃ½:", WAV_FILE)

if not WAV_FILE.exists():
    raise FileNotFoundError("ChÆ°a cÃ³ audio/input.wav. HÃ£y cháº¡y CELL 01 trÆ°á»›c.")
```

CELL 03 â€” Äá»c WAV vÃ  chuáº©n hÃ³a

```python
# CELL 03 - Äá»c file WAV

Fs, data = wavfile.read(WAV_FILE)

print("===== THÃ”NG TIN Dá»® LIá»†U =====")
print("Sampling rate:", Fs, "Hz")
print("Dtype:", data.dtype)
print("Shape:", data.shape)

# XÃ¡c Ä‘á»‹nh sá»‘ kÃªnh
if data.ndim == 1:
    channels = 1
    print("Channels: Mono")
else:
    channels = data.shape[1]
    print("Channels:", channels)

# Duration
n_samples = len(data)
duration = n_samples / Fs

print("Duration:", duration, "seconds")

# Sample width
sample_width = data.dtype.itemsize

print("Sample width:", sample_width, "bytes")
print("Bit depth:", sample_width * 8, "bits")

# File size
file_size = WAV_FILE.stat().st_size
print("File size:", file_size, "bytes")
print("File size:", file_size / 1024, "KB")

# Chuyá»ƒn sang float
data_float = data.astype(np.float64)

# Normalize vá» [-1, 1]
if np.issubdtype(data.dtype, np.integer):
    info = np.iinfo(data.dtype)
    scale = max(abs(info.min), abs(info.max))
    data_norm = data_float / scale
else:
    max_value = np.max(np.abs(data_float))
    if max_value > 1:
        data_norm = data_float / max_value
    else:
        data_norm = data_float

print("\nNormalized range:")
print("Min =", np.min(data_norm))
print("Max =", np.max(data_norm))
```

CELL B3 â€” Tá»± Ä‘á»™ng tÃ¬m 2 Ä‘oáº¡n khÃ¡c nhau

Äá»ƒ trÃ¡nh tá»± chá»n hai Ä‘oáº¡n giá»‘ng nhau, code sáº½ tÃ¬m:

má»™t Ä‘oáº¡n RMS tháº¥p
má»™t Ä‘oáº¡n RMS cao

```python
# B3 - Chá»n 2 Ä‘oáº¡n cÃ³ Ä‘áº·c tÃ­nh khÃ¡c nhau

segment_duration = 0.5
segment_samples = int(segment_duration * Fs)

# TÃ­nh RMS cá»§a tá»«ng cá»­a sá»• 0.5 s
rms_values = []
starts = []

step = segment_samples

for start in range(0, len(mono) - segment_samples, step):
    segment = mono[start:start + segment_samples]

    segment_rms = np.sqrt(np.mean(segment ** 2))

    rms_values.append(segment_rms)
    starts.append(start)

rms_values = np.array(rms_values)
starts = np.array(starts)

# Äoáº¡n RMS tháº¥p nháº¥t
low_idx = np.argmin(rms_values)

# Äoáº¡n RMS cao nháº¥t
high_idx = np.argmax(rms_values)

low_start = starts[low_idx]
high_start = starts[high_idx]

segment_low = mono[low_start:low_start + segment_samples]
segment_high = mono[high_start:high_start + segment_samples]

print("===== 2 ÄOáº N ÄÆ¯á»¢C CHá»ŒN =====")

print(
    f"Segment LOW  : "
    f"{low_start/Fs:.2f}s - {(low_start+segment_samples)/Fs:.2f}s"
)
print(f"RMS LOW       : {rms_values[low_idx]:.6f}")

print(
    f"Segment HIGH : "
    f"{high_start/Fs:.2f}s - {(high_start+segment_samples)/Fs:.2f}s"
)
print(f"RMS HIGH      : {rms_values[high_idx]:.6f}")
```

PDF yÃªu cáº§u:

CELL C1 â€” Chá»n Ä‘oáº¡n tÆ°Æ¡ng Ä‘á»‘i á»•n Ä‘á»‹nh 0.75 s

```python
# C1 - Tá»± tÃ¬m Ä‘oáº¡n tÆ°Æ¡ng Ä‘á»‘i á»•n Ä‘á»‹nh Ä‘á»ƒ FFT

fft_duration = 0.75
fft_samples = int(fft_duration * Fs)

# XÃ©t cÃ¡c Ä‘oáº¡n 0.75 s
candidate_starts = range(
    0,
    len(mono) - fft_samples,
    int(0.25 * Fs)
)

scores = []

for start in candidate_starts:

    segment = mono[start:start + fft_samples]

    # Chia thÃ nh nhiá»u frame nhá»
    frame_length = int(0.125 * Fs)

    rms_list = []

    for i in range(
        0,
        len(segment) - frame_length + 1,
        frame_length
    ):
        frame = segment[i:i + frame_length]

        frame_rms = np.sqrt(np.mean(frame ** 2))
        rms_list.append(frame_rms)

    rms_list = np.array(rms_list)

    mean_rms = np.mean(rms_list)

    if mean_rms > 1e-5:
        stability = np.std(rms_list) / mean_rms
    else:
        stability = np.inf

    scores.append((start, stability))

# Loáº¡i nhá»¯ng Ä‘oáº¡n quÃ¡ im láº·ng
rms_threshold = np.percentile(
    [s[1] for s in scores if np.isfinite(s[1])],
    70
)

valid_scores = [
    item for item in scores
    if np.isfinite(item[1])
]

# Chá»n Ä‘oáº¡n cÃ³ stability tháº¥p nháº¥t
fft_start, stability_score = min(
    valid_scores,
    key=lambda x: x[1]
)

fft_segment = mono[
    fft_start:
    fft_start + fft_samples
]

print("===== FFT SEGMENT =====")
print(f"Start      : {fft_start/Fs:.3f} s")
print(f"End        : {(fft_start+fft_samples)/Fs:.3f} s")
print(f"Duration   : {fft_duration:.3f} s")
print(f"Stability  : {stability_score:.6f}")
```

CELL C2 — FFT với cửa sổ Hamming, phổ biên độ và 3 đỉnh phổ

```python
# C2 - FFT với cửa sổ Hamming, vẽ phổ biên độ theo Hz và dB

window = np.hamming(len(fft_segment))
fft_windowed = fft_segment * window

nfft_base = 2 ** int(np.ceil(np.log2(len(fft_segment))))
nfft_values = [nfft_base, nfft_base * 2]

peak_summary = []

for nfft in nfft_values:
    spectrum = np.fft.rfft(fft_windowed, n=nfft)
    freqs = np.fft.rfftfreq(nfft, d=1 / Fs)

    magnitude = np.abs(spectrum)
    magnitude_db = 20 * np.log10(magnitude + 1e-12)

    peaks, _ = signal.find_peaks(
        magnitude_db,
        distance=20,
        prominence=10
    )

    if len(peaks) >= 3:
        top_peaks = peaks[np.argsort(magnitude_db[peaks])[-3:]]
    else:
        top_peaks = np.argsort(magnitude_db)[-3:]

    top_peaks = top_peaks[np.argsort(freqs[top_peaks])]
    peak_summary.append((nfft, freqs[top_peaks], magnitude_db[top_peaks]))

    plt.figure(figsize=(14, 5))
    plt.plot(freqs, magnitude_db)
    plt.scatter(freqs[top_peaks], magnitude_db[top_peaks], color="red")

    for p in top_peaks:
        plt.text(freqs[p], magnitude_db[p], f"{freqs[p]:.1f} Hz")

    plt.xlim(0, min(8000, Fs / 2))
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Magnitude (dB)")
    plt.title(f"C - Magnitude Spectrum, NFFT = {nfft}")
    plt.grid(True, alpha=0.3)
    plt.savefig(FIGURE_DIR / f"C_fft_nfft_{nfft}.png", dpi=200)
    plt.show()

    print(f"===== NFFT = {nfft} =====")
    print(f"Frequency-bin spacing Δf = Fs/NFFT = {Fs/nfft:.4f} Hz")
    print("Top 3 spectral peaks:")
    for p in top_peaks:
        print(f"- {freqs[p]:.2f} Hz, {magnitude_db[p]:.2f} dB")
```

D. STFT và Spectrogram

```python
# D1 - Spectrogram chuẩn: frame khoảng 25 ms, hop khoảng 10 ms

frame_ms = 25
hop_ms = 10

nperseg = int(round(frame_ms / 1000 * Fs))
hop_samples = int(round(hop_ms / 1000 * Fs))
noverlap = nperseg - hop_samples

# nfft phải >= nperseg để tránh lỗi ValueError trong signal.spectrogram.
nfft = 2 ** int(np.ceil(np.log2(nperseg)))

f, t_spec, S = signal.spectrogram(
    mono,
    fs=Fs,
    window="hamming",
    nperseg=nperseg,
    noverlap=noverlap,
    nfft=nfft,
    mode="magnitude"
)

S_db = 20 * np.log10(S + 1e-12)

plt.figure(figsize=(14, 5))
plt.pcolormesh(t_spec, f, S_db, shading="gouraud", cmap="magma")
plt.ylim(0, min(8000, Fs / 2))
plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")
plt.title("D - Spectrogram chuẩn: frame 25 ms, hop 10 ms")
plt.colorbar(label="Magnitude (dB)")
plt.savefig(FIGURE_DIR / "D_spectrogram_25ms.png", dpi=200)
plt.show()

print(f"Frame samples: {nperseg}")
print(f"Hop samples  : {hop_samples}")
print(f"NFFT         : {nfft}")
```

```python
# D2 - So sánh spectrogram với 3 độ dài frame: 10 ms, 25 ms, 50 ms

frame_list_ms = [10, 25, 50]
hop_ms = 10

plt.figure(figsize=(14, 12))

for idx, frame_ms in enumerate(frame_list_ms, 1):
    nperseg = int(round(frame_ms / 1000 * Fs))
    hop_samples = int(round(hop_ms / 1000 * Fs))
    noverlap = max(0, nperseg - hop_samples)
    nfft = 2 ** int(np.ceil(np.log2(nperseg)))

    f, t_spec, S = signal.spectrogram(
        mono,
        fs=Fs,
        window="hamming",
        nperseg=nperseg,
        noverlap=noverlap,
        nfft=nfft,
        mode="magnitude"
    )

    S_db = 20 * np.log10(S + 1e-12)

    plt.subplot(3, 1, idx)
    plt.pcolormesh(t_spec, f, S_db, shading="gouraud", cmap="magma")
    plt.ylim(0, min(8000, Fs / 2))
    plt.ylabel("Frequency (Hz)")
    plt.title(f"Frame = {frame_ms} ms")

plt.xlabel("Time (s)")
plt.tight_layout()
plt.savefig(FIGURE_DIR / "D_compare_frame_lengths.png", dpi=200)
plt.show()
```

E. Thí nghiệm cửa sổ

```python
# E - So sánh Rectangular và Hamming trên cùng một khung tín hiệu

frame_duration = 0.05
frame_samples = int(frame_duration * Fs)
frame = fft_segment[:frame_samples]

rect_window = np.ones(frame_samples)
hamming_window = np.hamming(frame_samples)

nfft = 2 ** int(np.ceil(np.log2(frame_samples * 8)))
freqs = np.fft.rfftfreq(nfft, d=1 / Fs)

rect_spec = np.abs(np.fft.rfft(frame * rect_window, n=nfft))
hamming_spec = np.abs(np.fft.rfft(frame * hamming_window, n=nfft))

rect_db = 20 * np.log10(rect_spec / np.max(rect_spec) + 1e-12)
hamming_db = 20 * np.log10(hamming_spec / np.max(hamming_spec) + 1e-12)

plt.figure(figsize=(14, 5))
plt.plot(freqs, rect_db, label="Rectangular")
plt.plot(freqs, hamming_db, label="Hamming")
plt.xlim(0, min(8000, Fs / 2))
plt.ylim(-100, 5)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Normalized magnitude (dB)")
plt.title("E - Log-spectrum: Rectangular vs Hamming")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig(FIGURE_DIR / "E_window_comparison.png", dpi=200)
plt.show()

print("===== NHẬN XÉT CỬA SỔ =====")
print("Rectangular có main-lobe hẹp hơn nhưng side-lobe cao hơn, nên dễ gây spectral leakage.")
print("Hamming có side-lobe thấp hơn nên giảm rò rỉ phổ, nhưng main-lobe rộng hơn.")
```

F. Lọc số

```python
# F1 - Thiết kế FIR Low-pass và High-pass, vẽ đáp ứng tần số H(f)

nyquist = Fs / 2
numtaps = 801

low_cutoff = min(4000, nyquist * 0.8)
high_cutoff = min(1000, nyquist * 0.4)

fir_low = signal.firwin(numtaps, low_cutoff, fs=Fs, pass_zero="lowpass")
fir_high = signal.firwin(numtaps, high_cutoff, fs=Fs, pass_zero="highpass")

w_low, h_low = signal.freqz(fir_low, worN=4096, fs=Fs)
w_high, h_high = signal.freqz(fir_high, worN=4096, fs=Fs)

plt.figure(figsize=(14, 5))
plt.plot(w_low, 20 * np.log10(np.abs(h_low) + 1e-12), label="Low-pass")
plt.plot(w_high, 20 * np.log10(np.abs(h_high) + 1e-12), label="High-pass")
plt.xlim(0, min(10000, nyquist))
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.title("F - Đáp ứng tần số FIR H(f)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig(FIGURE_DIR / "F_filter_response.png", dpi=200)
plt.show()

mono_low = signal.lfilter(fir_low, 1.0, mono)
mono_high = signal.lfilter(fir_high, 1.0, mono)

sf.write(AUDIO_DIR / "filtered_lowpass.wav", mono_low.astype(np.float32), Fs)
sf.write(AUDIO_DIR / "filtered_highpass.wav", mono_high.astype(np.float32), Fs)

print("Đã xuất file lọc:")
print(AUDIO_DIR / "filtered_lowpass.wav")
print(AUDIO_DIR / "filtered_highpass.wav")
```

```python
# F2 - So sánh phổ tín hiệu trước và sau khi lọc

def plot_spectrum_compare(original, filtered, title, filename):
    nfft = 2 ** int(np.ceil(np.log2(len(original))))
    freqs = np.fft.rfftfreq(nfft, d=1 / Fs)

    original_spec = np.abs(np.fft.rfft(original, n=nfft))
    filtered_spec = np.abs(np.fft.rfft(filtered, n=nfft))

    original_db = 20 * np.log10(original_spec + 1e-12)
    filtered_db = 20 * np.log10(filtered_spec + 1e-12)

    plt.figure(figsize=(14, 5))
    plt.plot(freqs, original_db, label="Before")
    plt.plot(freqs, filtered_db, label="After")
    plt.xlim(0, min(10000, Fs / 2))
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Magnitude (dB)")
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.savefig(FIGURE_DIR / filename, dpi=200)
    plt.show()

plot_spectrum_compare(mono, mono_low, "F - Spectrum before/after Low-pass", "F_lowpass_spectrum.png")
plot_spectrum_compare(mono, mono_high, "F - Spectrum before/after High-pass", "F_highpass_spectrum.png")
```

G. Lượng tử hóa, Resampling và Mã hóa

```python
# G1 - Lượng tử hóa tín hiệu ở 4, 8, 16 bit và tính SNR đo được

def quantize_signal(x, bits):
    levels = 2 ** bits
    x_clipped = np.clip(x, -1, 1)
    q = np.round((x_clipped + 1) / 2 * (levels - 1))
    x_quantized = q / (levels - 1) * 2 - 1
    return x_quantized


def calculate_snr(original, processed):
    noise = original - processed
    signal_power = np.mean(original ** 2)
    noise_power = np.mean(noise ** 2)
    return 10 * np.log10(signal_power / (noise_power + 1e-12))


quantization_results = []

for bits in [4, 8, 16]:
    mono_q = quantize_signal(mono, bits)
    snr = calculate_snr(mono, mono_q)
    output_quantized = AUDIO_DIR / f"quantized_{bits}bit.wav"

    sf.write(output_quantized, mono_q.astype(np.float32), Fs)

    quantization_results.append({
        "Bits": bits,
        "Levels": 2 ** bits,
        "SNR (dB)": snr,
        "Output file": str(output_quantized)
    })

quantization_table = pd.DataFrame(quantization_results)

print("===== BẢNG SNR SAU LƯỢNG TỬ HÓA =====")
display(quantization_table)

plt.figure(figsize=(8, 4))
plt.bar(
    quantization_table["Bits"].astype(str),
    quantization_table["SNR (dB)"],
    color=["#d95f02", "#1b9e77", "#7570b3"]
)
plt.xlabel("Số bit lượng tử hóa")
plt.ylabel("SNR đo được (dB)")
plt.title("G - SNR theo số bit lượng tử hóa")
plt.grid(axis="y", alpha=0.3)
plt.savefig(FIGURE_DIR / "G_quantization_snr.png", dpi=200)
plt.show()
```

```python
# G2 - Resample âm thanh về 16 kHz và 8 kHz, so sánh phổ và nghe thử

target_rates = [16000, 8000]

for target_fs in target_rates:
    num_samples = int(len(mono) * target_fs / Fs)
    mono_resampled = signal.resample(mono, num_samples)

    output_resampled = AUDIO_DIR / f"resampled_{target_fs}Hz.wav"
    sf.write(output_resampled, mono_resampled.astype(np.float32), target_fs)

    print(f"Đã resample về {target_fs} Hz:", output_resampled)

    nfft = 2 ** int(np.ceil(np.log2(len(mono_resampled))))
    freqs = np.fft.rfftfreq(nfft, d=1 / target_fs)
    spec = np.abs(np.fft.rfft(mono_resampled, n=nfft))
    spec_db = 20 * np.log10(spec + 1e-12)

    plt.figure(figsize=(14, 4))
    plt.plot(freqs, spec_db)
    plt.xlim(0, target_fs / 2)
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Magnitude (dB)")
    plt.title(f"G - Spectrum after resampling to {target_fs} Hz")
    plt.grid(True, alpha=0.3)
    plt.savefig(FIGURE_DIR / f"G_resampled_{target_fs}Hz_spectrum.png", dpi=200)
    plt.show()

    display(Audio(mono_resampled, rate=target_fs))
```
